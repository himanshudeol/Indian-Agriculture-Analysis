```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('ICRISAT-District Level Data.csv')
df.info()
df.head()
```

#### The dataset "ICRISAT-District Level Data.csv" contains agricultural data at the district level, covering multiple states, years, seasons, and crops in India.
#### 1. General Information: These columns provide contextual details about the data.
- State Name – The name of the state where the data was recorded.
- Dist Name – The name of the district within the state.
- Year – The year the data corresponds to.
- Season – The cropping season, such as Kharif, Rabi, or Annual.
#### 2. Crop Area (in 1000 ha): These columns represent the total area (land) allocated for different crops.
- RICE AREA (1000 ha) – Total land used for rice cultivation in 1000 hectares.
- WHEAT AREA (1000 ha) – Total land used for wheat cultivation.
- MAIZE AREA (1000 ha) – Total land used for maize cultivation.
- CHICKPEA AREA (1000 ha) – Total land used for chickpea cultivation.
- SORGHUM AREA (1000 ha) – Total land used for sorghum cultivation.
- VEGETABLE AREA (1000 ha) – Total land allocated for vegetable cultivation.
- KHARIF SORGHUM AREA (1000 ha) – Land allocated for sorghum in the Kharif season.
- RABI SORGHUM AREA (1000 ha) – Land allocated for sorghum in the Rabi season.
- Similar columns exist for other crops
#### 3. Crop Production Columns (in 1000 tons): These columns represent the total production output for different crops.
- RICE PRODUCTION (1000 tons) – Total rice production in 1000 tons.
- WHEAT PRODUCTION (1000 tons) – Total wheat production.
- MAIZE PRODUCTION (1000 tons) – Total maize production.
- CHICKPEA PRODUCTION (1000 tons) – Total chickpea production.
- SORGHUM PRODUCTION (1000 tons) – Total sorghum production.
- Similar columns exist for other crops
#### 4. Crop Yield Columns (Kg per ha): These columns represent the average yield (productivity) of different crops in kilograms per hectare.
- RICE YIELD (Kg per ha) – Rice yield per hectare.
- WHEAT YIELD (Kg per ha) – Wheat yield per hectare.
- MAIZE YIELD (Kg per ha) – Maize yield per hectare.
- CHICKPEA YIELD (Kg per ha) – Chickpea yield per hectare.
- SORGHUM YIELD (Kg per ha) – Sorghum yield per hectare.
- Similar columns exist for other crops

```python
#1. Crop Area Distribution
crops = ['RICE AREA (1000 ha)', 'WHEAT AREA (1000 ha)', 'MAIZE AREA (1000 ha)']
crop_areas = df[crops].sum()
print("The total area allocated to:")
print(crop_areas)
plt.figure(figsize=(8, 5))
crop_areas.plot(kind='bar', color=['blue', 'green', 'orange'])
plt.xlabel('Crops')
plt.ylabel('Total Area (1000 ha)')
plt.title('Crop Area Distribution')
plt.show()
```

```python
#2. Yearly Rice Production
rice_production = df.groupby('Year')['RICE PRODUCTION (1000 tons)'].sum()
print(f"Yearly Rice Production (1000 tons):\n{rice_production}")
max_production_year = rice_production.loc[rice_production == rice_production.max()].index[0]
print(f"The year with the highest rice production was: {max_production_year} with {rice_production.max() * 1000} tons")
highest_production_value = rice_production.max()
plt.figure(figsize=(10, 5))
plt.plot(rice_production.index, rice_production.values, marker='o', linestyle='-', color='orange')
plt.xlabel('Year')
plt.ylabel('Total Rice Production (1000 tons)')
plt.title('Yearly Rice Production')
plt.grid(True)
plt.show()
```

```python
#3. State-wise Wheat Production
wheat_production = df.groupby('State Name')['WHEAT PRODUCTION (1000 tons)'].sum().sort_values()
state_highest_production = wheat_production.idxmax()
state_lowest_production = wheat_production.idxmin()
print(f"State-wise Wheat Production (1000 tons):\n{wheat_production}")
print(f"The state with the highest wheat production is: {state_highest_production} with {wheat_production.max() * 1000} tons.")
print(f"The state with the lowest wheat production is: {state_lowest_production} with {wheat_production.min() * 1000} tons.")
plt.figure(figsize=(10, 6))
wheat_production.plot(kind='barh', color='teal')
plt.xlabel('Total Wheat Production (1000 tons)')
plt.ylabel('State')
plt.title('Wheat Production by State')
plt.show()
```

```python
#4. Crop Yields for Sorghum Yield
average_sorghum_yield = df['SORGHUM YIELD (Kg per ha)'].mean()
print(f"The average sorghum yield is: {average_sorghum_yield:.2f} Kg per ha")
plt.figure(figsize=(6, 5))
sns.boxplot(y=df['SORGHUM YIELD (Kg per ha)'], color='purple')
plt.ylabel('Sorghum Yield (Kg per ha)')
plt.title('Distribution of Sorghum Yields')
plt.show()
```

```python
#5. State-wise Vegetable Area
vegetable_area = df.groupby('State Name')['VEGETABLES AREA (1000 ha)'].sum()
total_vegetable_area = vegetable_area.sum()
state_max_area = vegetable_area.idxmax()
max_area_value = vegetable_area.max()
print(f"Total vegetable area: {total_vegetable_area * 1000} ha")
print(f"State with the maximum vegetable area: {state_max_area} with {max_area_value * 1000} ha")
plt.figure(figsize=(8, 8))
vegetable_area.plot(kind='pie', autopct='%1.1f%%', startangle=140)
plt.ylabel('')
plt.title('Vegetable Area Distribution by State')
plt.show()
```

### 2. Diversity of Crops
- Analysis: Count the number of different crops produced in each state.
- Visualization: Bar Chart
    * How: Create a bar chart showing the number of crops produced per state, with states on the x-axis and the count on the y-axis.

```python
#1. Area vs. Production
correlation = df['CHICKPEA AREA (1000 ha)'].corr(df['CHICKPEA PRODUCTION (1000 tons)'])
print(f"Correlation Between Chickpea Area and Production: {correlation:.2f}")
plt.figure(figsize=(7, 5))
sns.regplot(x=df['CHICKPEA AREA (1000 ha)'], y=df['CHICKPEA PRODUCTION (1000 tons)'], scatter_kws={'alpha':0.5}, line_kws={'color':'red'})
plt.xlabel('Chickpea Area (1000 ha)')
plt.ylabel('Chickpea Production (1000 tons)')
plt.title('Chickpea Area vs Production')
plt.grid(True)
plt.show()
```

```python
#2. Diversity of Crops
crop_columns = [col for col in df.columns if 'AREA (1000 ha)' in col]
crop_counts = (df[crop_columns] > 0).groupby(df['State Name']).sum().sum(axis=1)
plt.figure(figsize=(10, 6))
crop_counts.plot(kind='bar', color='cyan')
plt.xlabel('State')
plt.ylabel('Number of Crops')
plt.title('Diversity of Crops by State')
plt.xticks(rotation=90)
plt.show()
print("Number of different crops produced in each state:")
print(crop_counts)
print(f"\nMost diverse state: {crop_counts.idxmax()} with {crop_counts.max()} crops")
print(f"Least diverse state: {crop_counts.idxmin()} with {crop_counts.min()} crops")
```

## Hard Questions:
### 1. Longitudinal Yield Trends
- Analysis: Analyze yield changes for major pulses over the years.
- Visualization: Line Chart
    * How: Create a line chart for each pulse crop with years on the x-axis and yield on the y-axis, allowing for easy comparison of trends.
### 2. Sorghum Production Patterns
- Analysis: Identify planting patterns for kharif and rabi sorghum.
- Visualization: Heatmap
    * How: Create a heatmap showing the area planted for kharif and rabi sorghum across different districts, allowing for visual pattern recognition.

```python
#1. Longitudinal Yield Trends
pulses = ['CHICKPEA YIELD (Kg per ha)', 'PIGEONPEA YIELD (Kg per ha)', 'MINOR PULSES YIELD (Kg per ha)']
plt.figure(figsize=(10, 5))
for pulse in pulses:
    yield_trend = df.groupby('Year')[pulse].mean()
    plt.plot(yield_trend.index, yield_trend.values, marker='o', linestyle='-', label=pulse.split(' ')[0])
plt.xlabel('Year')
plt.ylabel('Yield (Kg per ha)')
plt.title('Longitudinal Yield Trends for Major Pulses')
plt.legend()
plt.grid(True)
plt.show()
```

```python
for pulse in pulses:
    yield_trend = df.groupby('Year')[pulse].mean()
    percentage_change = (yield_trend.pct_change() * 100).round(2)
    print(f"\n{pulse.split(' ')[0]} Pulses Yield Change (%):")
    print(percentage_change)
    highest_yield_year = yield_trend.idxmax()
    highest_yield_value = yield_trend.max()
    print(f"Year with highest {pulse.split(' ')[0]} pulses yield: {highest_yield_year} ({highest_yield_value:.2f} Kg/ha)")
    lowest_yield_year = yield_trend.idxmin()
    lowest_yield_value = yield_trend.min()
    print(f"Year with lowest {pulse.split(' ')[0]} pulses yield: {lowest_yield_year} ({lowest_yield_value:.2f} Kg/ha)")
```

```python
#9. Sorghum Production Patterns for Kharif & Rabi Sorghum
kharif_rabi = df.pivot_table(index='Dist Name', values=['KHARIF SORGHUM AREA (1000 ha)', 'RABI SORGHUM AREA (1000 ha)'], aggfunc='sum')
plt.figure(figsize=(10, 8))
sns.heatmap(kharif_rabi, cmap='Blues')
plt.xlabel('Season')
plt.ylabel('District')
plt.title('Sorghum Production Patterns')
plt.show()
```

```python
kharif_area = kharif_rabi['KHARIF SORGHUM AREA (1000 ha)']
rabi_area = kharif_rabi['RABI SORGHUM AREA (1000 ha)']
total_kharif_area = kharif_area.sum()
total_rabi_area = rabi_area.sum()
total_area = total_kharif_area + total_rabi_area
percent_kharif = (total_kharif_area / total_area) * 100
percent_rabi = (total_rabi_area / total_area) * 100
print(f"\nOverall, {percent_kharif:.1f}% of sorghum area is Kharif({(total_kharif_area * 1000):.1f} ha) and {percent_rabi:.1f}% is Rabi({(total_rabi_area * 1000):.1f} ha).")
```

```python
df['KHARIF SORGHUM AREA (1000 ha)'] = df['KHARIF SORGHUM AREA (1000 ha)'].apply(lambda x: np.nan if x < 0 else x)
df_cleaned = df.dropna(subset=['KHARIF SORGHUM AREA (1000 ha)']) 
kharif_by_district = df_cleaned.groupby('Dist Name')['KHARIF SORGHUM AREA (1000 ha)'].sum()
print(kharif_by_district)
print(f"\nMost dominant district for Kharif sorghum: {kharif_by_district.idxmax()} ({(kharif_by_district.max() * 1000):.1f} ha)")
print(f"\nMinimum dominant district for Kharif sorghum: {kharif_by_district.idxmin()} ({(kharif_by_district.min() * 1000):.1f} ha)")
```

```python
df['RABI SORGHUM AREA (1000 ha)'] = df['RABI SORGHUM AREA (1000 ha)'].apply(lambda x: np.nan if x < 0 else x)
df_cleaned = df.dropna(subset=['RABI SORGHUM AREA (1000 ha)']) 
rabi_by_district = df_cleaned.groupby('Dist Name')['RABI SORGHUM AREA (1000 ha)'].sum()
print(rabi_by_district)
print(f"\nMost dominant district for Rabi sorghum: {rabi_by_district.idxmax()} ({(rabi_by_district.max() * 1000):.1f} ha)")
print(f"\nMinimum dominant district for Rabi sorghum: {rabi_by_district.idxmin()} ({(rabi_by_district.min() * 1000):.1f} ha)")
```
